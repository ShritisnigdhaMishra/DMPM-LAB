Write a Python code using sklearn library to do Decision tree operation or induction on iris dataset

```python
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier
df=sns.load_dataset('iris')
x=df[['petal_length','petal_width']]
y=df['species']
clf=DecisionTreeClassifier(criterion='entropy',max_depth=1,random_state=42)
clf.fit(x,y)
root_node=0
best_feature_idx=clf.tree_.feature[root_node]
best_feature_name=x.columns[best_feature_idx]
best_threshold=round(clf.tree_.threshold[root_node],2)
parent_entropy=clf.tree_.impurity[root_node]
left_child=clf.tree_.children_left[root_node]
right_child=clf.tree_.children_right[root_node]
n_total=clf.tree_.n_node_samples[root_node]
n_left=clf.tree_.n_node_samples[left_child]
n_right=clf.tree_.n_node_samples[right_child]
remainder_entropy=((n_left/n_total)*clf.tree_.impurity[left_child]+(n_right/n_total)*clf.tree_.impurity[right_child])
info_gain=round(parent_entropy-remainder_entropy,4)
print(f"The best split chosen by the algorithm is:[{best_feature_name}<={best_threshold}]")
print(f"Maximized information gain achieved:{info_gain}")
```
