1.Find the frequent items using association mining rule in market basket analysis where there is 5 transaction id items are in T1 bread,milk in T2 bread,diaper,beer,eggs in T3 milk,diaper,beer,coke in T4 bread,milk,diaper,beer in T5 bread,milk,diaper,coke
Minimum support=60%


```python
import pandas as pd
def reco(data,min_support):
    trans=[set(t) for t in data.values()]
    n=len(trans)
    min_count=min_support*n
    unique_items=sorted(list(set(item for t in trans for item in t)))
    curr_cand=[frozenset([item])for item in unique_items]
    records=[]
    while curr_cand:
        freq_k=[]
        for cand in curr_cand:
          count=sum(1 for t in trans if cand.issubset(t))
          if count>=min_count:
            support=count/n
            freq_k.append(cand)
            records.append({"Itemset Size":len(cand),"Support":support,"Frequent Itemset":f"{{{','.join(sorted(list(cand)))}}}"})
        if not freq_k:
          break
        next_cand=set()
        for i in range(len(freq_k)):
          for j in range(i+1,len(freq_k)):
            itemset1=sorted(list(freq_k[i]))
            itemset2=sorted(list(freq_k[j]))
            if itemset1[:-1]==itemset2[:-1]:
              next_cand.add(freq_k[i] | freq_k[j])
        curr_cand=list(next_cand)
    df=pd.DataFrame(records)
    df["support"]=(df["Support"]*100).astype(int).astype(str)+"%"
    df=df.sort_values(by=["Itemset Size","Frequent Itemset"]).reset_index(drop=True)
    return df
dataset={"T1":["Bread","Milk"],
         "T2":["Bread","Diaper","Beer","Eggs"],
         "T3":["Milk","Diaper","Beer","Coke"],
         "T4":["Bread","Milk","Diaper","Beer"],
         "T5":["Bread","Milk","Diaper","Coke"]}
min_support=0.6
res=reco(dataset,min_support)
print(res.to_string(index=False))
```

2.As per given dataset find the below questions and perform the required interpretation.
100 people were interviewed outside a chocolate shop to find out which flavour of chocolate cream they preferred.The results are given in the table below classified by gender
           Strawberry        Coffee       Orange        Vanila
 Male         23               18           8              8
 Female       15               6            12             10
We want to determine whether the flavour of chocolate cream is independent of gender.
i.State the null hypothesis and alternative hypothesis
ii.Calculate the degrees of freedom
iii.Create the expected frequency table for this data

```python
import pandas as pd
idx=["Strawberry","Coffee","Orange","Vanilla"]
data={"Male":[23,18,8,8],"Female":[15,6,12,10]}
df=pd.DataFrame(data,index=idx)
O=df.values.tolist()
R=len(O)
C=len(O[0])
r_tot=[sum(r) for r in O]
c_tot=[sum(O[i][j] for i in range(R)) for j in range(C)]
G=sum(r_tot)
E_mat=[]
x2=0
for r in range(R):
    e_row=[]
    for c in range(C):
        O_val=O[r][c]
        e_val=(r_tot[r]*c_tot[c])/G
        e_row.append(e_val)
        x2 +=((O_val-e_val)**2)/e_val
    E_mat.append(e_row)
df_E=pd.DataFrame(E_mat,index=idx,columns=data.keys())
df=(R-1)*(C-1)
print("\n--- a.Hypothesis ---")
print("Null Hypothesis (H0):Chocolate cream flavour preference is independent of gender")
print("Alternative Hypothesis (H1):Chocolate cream flavour preference is dependent on gender\n")
print("--- b.Degrees of freedom ---")
print(f"Degrees of freedom (df):{df}\n")
print("--- c.Expected frequency table ---")
print(df_E.round(2),"\n")
print("--- Full statistical summary ---")
print(f"Calculated Chi-square value (x2):{x2:.4f}")
tab_val=7.815
print(f"Tabulated Chi-square value (alpha=0.05):{tab_val:.4f}")
if x2<tab_val:
    print("\nConclusion:Accept H0.Flavour preference is independent of gender")
else:
    print("\nConclusion:Reject H0.Flavour preference is dependent on gender")
```
