4.Write a python program using seaborn and matplotlib to load the Titanic dataset,clean missing values and visualize numerical features using a boxplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
data=sns.load_dataset("titanic")
print("Initial row dataset:\n",data)
print("\n Null value counts:\n",data.isnull().sum())
data=data.dropna()
numerical_data=data.select_dtypes(include=[float,int])
plt.boxplot(numerical_data,tick_labels=numerical_data.columns)
plt.xticks(rotation=45)
plt.title("Boxplot of Numerical Features")
plt.show()
```
