```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('Iris.csv')
```

Use various data visualization tools or techniques to implement scatter plot


```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="SepalLengthCm",y="PetalLengthCm",hue="Species",alpha=0.8)
plt.title("Sepal Length vs Petal Length by Species")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.show()
```

2.Use various data visualization tools and techniques to implement line plot

```python
avgpetal=df.groupby("Species")["PetalLengthCm"].mean().reset_index()
plt.figure(figsize=(8,5))
sns.lineplot(data=avgpetal,x="Species",y="PetalLengthCm",marker="o")
plt.title("Average Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Petal Length (cm)")
plt.show()
```

Use various data visualization tools and techniques to implement bar chart

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df,x="Species",y="SepalLengthCm",errorbar=None)
plt.title("Average Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Sepal Length (cm)")
plt.show()
```

Use various data visualization tools and techniques to implement box plot

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="Species",y="PetalLengthCm")
plt.title("Distribution of Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")
plt.show()
```

Use various data visualization tools and techniques to implement pie chart

```python
speciescount=df["Species"].value_counts()
plt.figure(figsize=(7,7))
plt.pie(speciescount.values,labels=speciescount.index,autopct="%1.1f%%",startangle=90)
plt.title("Distribution of Iris Species")
plt.show()
```

Use various data visualization tools and techniques to implement histogram

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df,x="PetalLengthCm",bins=15,edgecolor="black",linewidth=1)
plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency (Number of Flowers)")
plt.show()
```
