1.WAP to implement z-score normalisation,min-max normalisation and decimal scaling normalization on 10,20,30,40,50

```python
import numpy as np
data=np.array([10,20,30,40,50])
mean=np.mean(data)
std=np.std(data)
z_score=(data-mean)/std
min=np.min(data)
max=np.max(data)
min_max=(data-min)/(max-min)
j=0
while max/(10**j)>=1:
    j=j+1
decimal=data/(10**j)
print("Original data:",data)
print("\nZ-Score normalization:")
print(z_score.round(3))
print("\nMin-Max normalization:")
print(min_max.round(3))
print("\nDecimal scaling normalization:")
print(decimal.round(3))
```

2.WAP to implement the decimal scaling normalization on a dataset from kaggle

```python
import numpy as np
import pandas as pd
df=pd.read_csv("Iris.csv")
columns=['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
for col in columns:
    max=df[col].max()
    j=0
    while max/(10**j)>=1:
        j=j+1
    df[col]=df[col]/(10 ** j)
    df[col]=round(df[col],3)
print("Decimal scaling normalized dataset:")
print(df.head())
```

3.WAP to implement min-max normalization on any dataset and display the normalized values

```python
import numpy as np
import pandas as pd
df=pd.read_csv("Iris.csv")
columns=['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
for col in columns:
    min=df[col].min()
    max=df[col].max()
    df[col]=(df[col]-min)/(max - min)
    df[col]=round(df[col],3)
print("Min-Max normalized dataset:")
print(df.head())
```

4.Implement the z-score normalization on any dataset from kaggle and display the mean,variance and standard deviation

```python
import numpy as np
import pandas as pd
df=pd.read_csv("Iris.csv")
columns=['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
for col in columns:
    mean=df[col].mean()
    var=df[col].var()
    std=df[col].std()
    print("\nColumn:",col)
    print("Mean:",round(mean,3))
    print("Variance:",round(var,3))
    print("Standard deviation:",round(std,3))
    df[col]=(df[col]-mean)/std
    df[col]=round(df[col],3)
print("\nZ-Score normalized dataset:")
print(df.head())
```
