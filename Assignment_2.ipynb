WAP to find the mean of data points without using functions

```python
data=[15,13,18,16,14,17,12]
mean=sum(data)/len(data)
print("Mean:",mean)
```

WAP to print the mean of data points with using functions

```python
from statistics import mean
data=[15,13,18,16,14,17,12]
mean=mean(data)
print("Mean:",mean)
```

WAP to find the median of data points without using functions

```python
data=[32,34,35,35,36,37,38,39,39,47]
data.sort()
n=len(data)
if n%2==0:
    median=(data[n//2-1]+data[n//2])/2
else:
    median=data[(n+1)//2]
print("Median=",median)
```

WAP to find the median of data points with using functions

```python
from statistics import median
data=[32,34,35,35,36,37,38,39,39,47]
median=median(data)
print("Median:",median)
```

WAP to find the mode of data points without using functions

```python
data=[5,5,5,5,6,6,6,7,8]
mode=max(set(data),key=data.count)
print("Mode:",mode)
```

WAP to find the median of data points with using functions

```python
from statistics import mode
data=[5,5,5,5,6,6,6,7,8]
mode=mode(data)
print("Mode:",mode)
```

WAP to print the variance of the data points without functions


```python
data=[15,13,18,16,14,17,12]
mean=sum(data)/len(data)
variance=sum((x-mean)**2 for x in data)/len(data)
print("Variance:",variance)
```

WAP to print the variance of the data points with functions

```python
from statistics import variance
data=[15,13,18,16,14,17,12]
variance=variance(data)
print("Variance:",variance)
```

WAP to print the standard deviation of the data points with or without functions

```python
data=[15,13,18,16,14,17,12]
mean=sum(data)/len(data)
variance=sum((x-mean)**2 for x in data)/len(data)
stddeviation=(variance)**0.5
print("Standard Deviation:",stddeviation)
```

```python
from statistics import stdev
data=[15,13,18,16,14,17,12]
stddeviation=stdev(data)
print("Standard Deviation:",stddeviation)
```

WAP to find the variance and standard deviation of the data points entered by the user


```python
from statistics import variance,stdev
data = list(map(float,input("Enter data(comma separated):").split(",")))
print("Variance:",variance(data))
print("Standard deviation:",stdev(data))
```
