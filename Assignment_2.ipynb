WAP to find the mean of data points without using functions

```python
data=[15,13,18,16,14,17,12]
mean=sum(data)/len(data)
print("Mean:",mean)
```

WAP to print the mean of data points with using functions

```python
from statistics import mean
data=[15,13,18,16,14,17,12]
mean=mean(data)
print("Mean:",mean)
```

WAP to find the median of data points without using functions

```python
data=[32,34,35,35,36,37,38,39,39,47]
data.sort()
n=len(data)
if (n%2==0):
    median=(data[n//2-1]+data[n//2])/2
else:
    median=data[(n+1)//2]
print("Median=",median)
```

WAP to find the median of data points with using functions

```python
from statistics import median
data=[32,34,35,35,36,37,38,39,39,47]
median=median(data)
print("Median:",median)
```

WAP to find the mode of data points without using functions

```python
data=[5,5,5,5,6,6,6,7,8]
mode=data[0]
maxcount=0
for i in data:
    count=0
    for j in data:
        if (i==j):
            count+= 1
    if (count>maxcount):
        maxcount=count
        mode=i
print("Mode:",mode)
```

WAP to find the mode of data points with using functions

```python
from statistics import mode
data=[5,5,5,5,6,6,6,7,8]
mode=mode(data)
print("Mode:",mode)
```

WAP to print the variance of the data points without using functions

```python
data=[15,13,18,16,14,17,12]
mean=sum(data)/len(data)
variance=sum((x-mean)**2 for x in data)/(len(data)-1)
print(f"Variance:{variance:.4f}")
```

WAP to print the variance of the data points with using functions

```python
from statistics import variance
data=[15,13,18,16,14,17,12]
variance=variance(data)
print(f"Variance:{variance:.4f}")
```

WAP to print the standard deviation of the data points with or without using functions

```python
from statistics import stdev
data=[15,13,18,16,14,17,12]
print(f"Standard deviation:{stdev(data):.4f}")
```

```python
data=[15,13,18,16,14,17,12]
mean=sum(data)/len(data)
variance=sum((x-mean)**2 for x in data)/(len(data)-1)
stddeviation=(variance)**0.5
print(f"Standard deviation:{stddeviation:.4f}")
```
