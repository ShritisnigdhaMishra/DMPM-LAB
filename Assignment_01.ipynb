WAP to find the largest of three numbers.

```python
a=int(input("Enter first number:"))
b=int(input("Enter second number:"))
c=int(input("Enter third number:"))
if(a>b and a>c):
    print(f"{a} is the largest")
elif(b>c):
    print(f"{b} is the largest")
else:
    print(f"{c} is the largest")
```

WAP to check whether a number is even or odd.

```python
num=int(input("Enter number:"))
if(num%2==0):
  print("It is an even number")
else:
  print("It is an odd number")
```

WAP to check whether a string is a palindrome.

```python
str=input("Enter a string:")
rev=str[::-1]
if(rev==str):
  print("The string is palindrome")
else:
  print("The string is not palindrome")
```

WAP to implement a simple calculator.

```python
a=int(input("Enter the first number:"))
b=int(input("Enter the second number:"))
print("1.Addition\n2.Subtraction\n3.Multiplication\n4.Division\n5.Modulo")
choice=int(input("Enter your choice (1-5):"))
if(choice==1):
    print("Addition:",(a+b))
elif(choice==2):
    print("Subtraction:",(a-b))
elif(choice==3):
    print("Multiplication:",(a*b))
elif(choice==4):
    if(b!=0):
        print("Division:",(a/b))
    else:
        print("Can't divided by zero")
elif(choice==5):
    print("Modulo:",(a%b))
else:
    print("Invalid choice")
```

WAP to count vowels in a string.

```python
s=input("Enter a string:")
vowels="aeiouAEIOU"
count=0
for char in s:
    if char in vowels:
        count +=1
print("Number of vowels in the string:",count)
```

WAP to check whether a number is prime.

```python
num=int(input("Enter a number:"))
for i in range(2,num):
    if(num%i==0):
        print("The number is not prime")
        break
else:
    if(num>1):
        print("The number is prime")
    else:
        print("The number is not prime")
```

WAP to find the sum of digits of a number.

```python
a=int(input("Enter a number:"))
sum=0
while(a>0):
    sum=sum+a%10
    a=a//10
print("Sum of digits:",sum)
```

WAP to reverse a number.

```python
num=int(input("Enter a number:"))
rev=0
while(num>0):
    rev=rev*10+num%10
    num=num//10
print("Reversed number:",rev)
```

WAP to find the factorial of a number.

```python
num=int(input("Enter a number:"))
fact=1
for i in range(1,num+1):
    fact=fact*i
print("Factorial is:",fact)
```
